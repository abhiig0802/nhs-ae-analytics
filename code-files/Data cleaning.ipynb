```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import glob
import os

```

```python
folder_path = Path("../data/raw")

csv_files = list(folder_path.glob("*.csv"))

dfs = []

for f in csv_files:
    temp = pd.read_csv(f, encoding='latin-1')
    temp['source_file'] = os.path.basename(f) 
    dfs.append(temp)
df = pd.concat(dfs, ignore_index=True)

print(f"Total rows loaded: {len(df)}")
print(f"Total columns: {len(df.columns)}")
print(f"Months loaded: {df['source_file'].nunique()}")
```

```python
cols_to_drop = []
for col in df.columns:
    if "Unnamed" in str(col):
        cols_to_drop.append(col)
    elif col == "a":
        cols_to_drop.append(col)
    elif col == "source_file":
        cols_to_drop.append(col)

df = df.drop(columns=cols_to_drop)
```

```python
df.head()
```

```python
df = df.rename(columns={
    'Period': 'period', 
    'Org Code': 'org_code',
    'Parent Org': 'parent_org',
    'Org name': 'org_name',
    'A&E attendances Type 1': 'att_type1',
    'A&E attendances Type 2': 'att_type2',
    'A&E attendances Other A&E Department': 'att_other',
    'A&E attendances Booked Appointments Type 1': 'att_booked_type1',
    'A&E attendances Booked Appointments Type 2': 'att_booked_type2',
    'A&E attendances Booked Appointments Other Department' : 'att_booked_other',
    'Attendances over 4hrs Type 1': 'over4hr_type1',
    'Attendances over 4hrs Type 2': 'over4hr_type2',
    'Attendances over 4hrs Other Department': 'over4hr_other',
    'Attendances over 4hrs Booked Appointments Type 1': 'over4hr_booked_type1',
    'Attendances over 4hrs Booked Appointments Type 2': 'over4hr_booked_type2',
    'Attendances over 4hrs Booked Appointments Other Department': 'over4hr_booked_other',
    'Patients who have waited 4-12 hs from DTA to admission' : 'wait_4_12hr',
    'Patients who have waited 12+ hrs from DTA to admission' : 'wait_12hr_plus',
    'Emergency admissions via A&E - Type 1': 'emerg_admit_type1',
    'Emergency admissions via A&E - Type 2': 'emerg_admit_type2',
    'Emergency admissions via A&E - Other A&E department' : 'emerg_admit_other',
    'Other emergency admissions': 'other_emerg_admit',
    'source_file': 'source_file'
})
print("Columns renamed!")
```

```python
# Strip whitespace 
df['org_name'] = df['org_name'].str.strip()

# Now remove only what actually exists in your data
remove_keywords = ['TOTAL', 'Total']

mask = df['org_name'].str.contains('|'.join(remove_keywords), na=False)
print(f"Summary rows found: {mask.sum()}")
print(f"Summary row names: {df[mask]['org_name'].unique()}")

df_clean = df[~mask].copy()
print(f"\nTrust rows remaining: {len(df_clean)}")
print(f"Unique trusts: {df_clean['org_name'].nunique()}")
```

```python
# Step 1 — Remove the 'MSitAE-' prefix
df_clean['period'] = df_clean['period'].str.replace('MSitAE-', '', regex=False)

# Step 2 — Convert to proper datetime
df_clean['period'] = pd.to_datetime(df_clean['period'], format='%B-%Y')

print("Period column fixed!")
print(f"\nSample values:")
print(df_clean['period'].unique())
print(f"\nDate range:")
print(f"Earliest: {df_clean['period'].min()}")
print(f"Latest:   {df_clean['period'].max()}")
```

```python
numeric_cols = [
    'att_type1', 'att_type2', 'att_other',
    'att_booked_type1', 'att_booked_type2', 'att_booked_other',
    'over4hr_type1', 'over4hr_type2', 'over4hr_other',
    'over4hr_booked_type1', 'over4hr_booked_type2', 'over4hr_booked_other',
    'wait_4_12hr', 'wait_12hr_plus',
    'emerg_admit_type1', 'emerg_admit_type2',
    'emerg_admit_other', 'other_emerg_admit'
]

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

print("Numeric columns fixed!")
print(df_clean[numeric_cols].dtypes)
```

```python
# Total attendances across all types
df_clean['total_attendances'] = (
    df_clean['att_type1'] +
    df_clean['att_type2'] +
    df_clean['att_other']
)

# Total breaches across all types
df_clean['total_over4hr'] = (
    df_clean['over4hr_type1'] +
    df_clean['over4hr_type2'] +
    df_clean['over4hr_other']
)

print(f"total_attendances created — sample values:")
print(df_clean['total_attendances'].head())

print(f"\ntotal_over4hr created — sample values:")
print(df_clean['total_over4hr'].head())
```

```python
# 4-hour performance percentage per trust per month
df_clean['pct_within_4hr'] = (
    (df_clean['total_attendances'] - df_clean['total_over4hr']) /
    df_clean['total_attendances'] * 100
).round(2)

print("pct_within_4hr created!")
print(f"\nMin performance: {df_clean['pct_within_4hr'].min()}%")
print(f"Max performance: {df_clean['pct_within_4hr'].max()}%")
print(f"Average performance: {df_clean['pct_within_4hr'].mean():.1f}%")
```

```python
df_clean['breach'] = (df_clean['pct_within_4hr'] < 95).astype(int)

print("=== TARGET VARIABLE CREATED ===")
print(f"Total rows: {len(df_clean)}")
print(f"Breached (1): {(df_clean['breach']==1).sum()}")
print(f"Met target (0): {(df_clean['breach']==0).sum()}")
print(f"Overall breach rate: {df_clean['breach'].mean()*100:.1f}%")

df_clean[['org_name','period','total_attendances','pct_within_4hr','breach']].head(10)
```

```python
before = len(df_clean)

# Remove rows where attendances are zero or percentage is NaN
df_clean = df_clean[df_clean['total_attendances'] > 0].copy()
df_clean = df_clean.dropna(subset=['pct_within_4hr'])

# Recreate breach column after removing bad rows
df_clean['breach'] = (df_clean['pct_within_4hr'] < 95).astype(int)

after = len(df_clean)
print(f"Rows removed: {before - after}")
print(f"Rows remaining: {after}")
print(f"Unique trusts: {df_clean['org_name'].nunique()}")
print(f"Breach rate: {df_clean['breach'].mean()*100:.1f}%")
```

```python
df_clean.to_csv("../data/cleaned/ae_clean.csv", index=False)
print("ae_clean.csv saved!")
print(f"Final shape: {df_clean.shape}")
```
